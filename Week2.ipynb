```python
def apply_label(img_input):
    h, w = img_input.shape[:2]
    bar_h = h // 12

    img_res = img_input.copy()
    img_res[h - bar_h:, :] = 0

    font = cv.FONT_HERSHEY_SIMPLEX
    text = "Dokumentasi PCVK - 2026"
    font_scale = h / 800
    thickness = max(1, int(h / 400))

    text_size = cv.getTextSize(text, font, font_scale, thickness)[0]
    text_x = int(w * 0.02)
    text_y = h - (bar_h - text_size[1]) // 2

    cv.putText(img_res, text, org=(text_x, text_y), fontFace=font, fontScale=font_scale, color=(255, 255, 255), thickness=thickness, lineType=cv.LINE_AA)

    return img_res

img_cs1 = apply_label(imgRGB)
plt.imshow(img_cs1)
plt.show()
```

```python
def make_thumbnail(img_input, store_name):
    h, w = img_input.shape[:2]
    side = max(h, w)

    canvas = np.full((side, side, 3), 255, dtype=np.uint8)

    y_off = (side - h) // 2
    x_off = (side - w) // 2

    canvas[y_off:y_off+h, x_off:x_off+w] = img_input

    font = cv.FONT_HERSHEY_SIMPLEX
    font_scale = side / 800
    thickness = max(1, int(side / 300))

    cv.putText(canvas, text=store_name, org=(int(side * 0.05), int(side * 0.1)), fontFace=font, fontScale=font_scale, color=(0, 0, 0), thickness=thickness, lineType=cv.LINE_AA)

    return canvas

img_cs2 = make_thumbnail(imgRGB, "Toko Zid")
plt.imshow(img_cs2)
plt.show()
```
