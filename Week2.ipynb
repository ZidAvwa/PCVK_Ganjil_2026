```python
black_img = np.zeros(shape=(512,512,3),dtype=np.int16)
plt.imshow(black_img)
```

```python
cv.rectangle(black_img,pt1=(200,200),pt2=(300,300),color=(0,0,255), thickness=15)
plt.imshow(black_img)
```

```python
font = cv.FONT_HERSHEY_SIMPLEX
cv.putText(black_img,text='Hello',org=(10,500),fontFace=font, fontScale=4,color=(255,255,0), thickness=2, lineType=cv.LINE_AA)
plt.imshow(black_img)
```

```python
import random

img_boxes = imgRGB.copy()
for _ in range(5):
    x1 = random.randint(0, img_boxes.shape[1] - 50)
    y1 = random.randint(0, img_boxes.shape[0] - 50)
    x2 = x1 + random.randint(10, 50)
    y2 = y1 + random.randint(10, 50)
    color = (random.randint(0, 255), random.randint(0, 255), random.randint(0, 255))
    cv.rectangle(img_boxes, pt1=(x1, y1), pt2=(x2, y2), color=color, thickness=3)

plt.imshow(img_boxes)
```
